```python
import pandas as pd
import glob
from tqdm import tqdm


file_paths = glob.glob("data/McDonald's_Indonesia/*.csv")

dfs = []

for file in tqdm(file_paths, desc="Loading CSV"):
    try:
        df = pd.read_csv(
            file,
            sep=';',                 
            engine='python',
            encoding='utf-8',
            quotechar='"',
            on_bad_lines='skip'
        )
        dfs.append(df)
    except Exception as e:
        print(f"[ERROR] {file}: {e}")

data = pd.concat(dfs, ignore_index=True)

print("Total rows:", len(data))

print(data.columns.tolist())
```

```python
# hapus tanda ' pada seluruh kolom bertipe string
for col in data.select_dtypes(include=['object']).columns:
    data[col] = data[col].str.replace("'", "", regex=False)

# tampilkan 5 baris pertama setelah pembersihan
print(data.head())
```

```python
# hapus kolom yang tidak diperlukan, yakni kolom setelah kolom "label"
cols_to_keep = data.columns.tolist()
if 'label' in cols_to_keep:
    label_index = cols_to_keep.index('label')
    cols_to_keep = cols_to_keep[:label_index + 1]
    data = data[cols_to_keep]
    print("Columns after dropping unnecessary ones:", data.columns.tolist())
    print("Total rows after dropping unnecessary columns:", len(data))
else:
    print("'label' column not found. No columns were dropped.")
```

```python
# omit duplicate rows
data = data.drop_duplicates()
print("Total rows after dropping duplicates:", len(data))
```

```python
# ganti separator ; dengan ,
data.to_csv('total_data.csv', index=False, sep=',', encoding='utf-8')

print("Combined cleaned CSV saved.")
```

```python
df = pd.read_csv(
	"total_data.csv",
	encoding='utf-8',
	sep=',',
	engine='python',
	on_bad_lines='skip'
)

print(df.shape)
print(df.columns)
```

```python
import re
import emoji

def clean_text(text):
    text = re.sub(r'http\S+|www\S+|https\S+', '', text, flags=re.MULTILINE)
    text = re.sub(r'@\w+', '', text)
    text = re.sub(r'#\w+', '', text)
    text = re.sub(r'\s+', ' ', text).strip()
    text = emoji.replace_emoji(text, replace='')
    return text

text_col = 'text' if 'text' in df.columns else 'content'
df['cleaned_text'] = df[text_col].fillna('').apply(clean_text)
print(df[[text_col, 'cleaned_text']].head())
print(len(df))
```

```python
# simpan data yang sudah diproses tanpa emoji ke file baru
df.to_csv('total_data_cleaned.csv', index=False, sep=',', encoding='utf-8')
print("Final cleaned CSV without emoji saved.")
```
