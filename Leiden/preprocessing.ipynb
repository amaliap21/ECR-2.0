```python
import pandas as pd
import glob
from tqdm import tqdm


file_paths = glob.glob("data/IGRS/*.csv")

dfs = []

for file in tqdm(file_paths, desc="Loading CSV"):
    try:
        df = pd.read_csv(
            file,
            sep=';',                 
            engine='python',
            encoding='utf-8',
            quotechar='"',
            on_bad_lines='skip'
        )
        dfs.append(df)
    except Exception as e:
        print(f"[ERROR] {file}: {e}")

data = pd.concat(dfs, ignore_index=True)

print("Total rows:", len(data))

print(data.columns.tolist())
```

```python
# hapus tanda ' pada seluruh kolom bertipe string
for col in data.select_dtypes(include=['object']).columns:
    data[col] = data[col].str.replace("'", "", regex=False)

# tampilkan 5 baris pertama setelah pembersihan
print(data.head())
```

```python
# hapus kolom yang tidak diperlukan, yakni kolom setelah kolom "label"
cols_to_keep = data.columns.tolist()
if 'label' in cols_to_keep:
    label_index = cols_to_keep.index('label')
    cols_to_keep = cols_to_keep[:label_index + 1]
    data = data[cols_to_keep]
    print("Columns after dropping unnecessary ones:", data.columns.tolist())
    print("Total rows after dropping unnecessary columns:", len(data))
else:
    print("'label' column not found. No columns were dropped.")
```

```python
# omit duplicate rows
data = data.drop_duplicates()
print("Total rows after dropping duplicates:", len(data))
```

```python
# ganti separator ; dengan ,
data.to_csv('total_data.csv', index=False, sep=',', encoding='utf-8')

print("Combined cleaned CSV saved.")
```

# Preprocessing Excel (xlsx) files

Pipeline untuk memproses file `.xlsx` (mis. data Komdigi). File xlsx besar dikonversi ke CSV per-file terlebih dahulu (pakai `python-calamine` agar cepat & tahan terhadap metadata `<dimension>` yang rusak), lalu digabung dan dibersihkan mengikuti alur yang sama dengan bagian IGRS.

```python
# install dependency (jalankan sekali saja)
%pip install -q python-calamine openpyxl
```

```python
# Konversi setiap xlsx → CSV (satu CSV per file xlsx, disimpan di folder yang sama)
import os
import glob
import pandas as pd
from tqdm import tqdm

# Ubah pola di bawah jika folder sumber berbeda
xlsx_paths = sorted(glob.glob("data/Komdigi_01-Jan-2026_15-May-2026_dxyNySNf1d/*.xlsx"))
print(f"Found {len(xlsx_paths)} xlsx files")

for f in tqdm(xlsx_paths, desc="XLSX → CSV"):
    out = os.path.splitext(f)[0] + ".csv"
    if os.path.exists(out):
        print(f"  [skip] {os.path.basename(out)} sudah ada")
        continue
    try:
        df = pd.read_excel(f, engine="calamine")
        df.to_csv(out, index=False, sep=",", encoding="utf-8")
        print(f"  [ok] {os.path.basename(out)}: {len(df)} rows, {len(df.columns)} cols")
    except Exception as e:
        print(f"  [ERROR] {f}: {e}")
```

```python
# Gabungkan seluruh CSV hasil konversi menjadi satu DataFrame `data`
csv_paths = sorted(glob.glob("data/Komdigi_01-Jan-2026_15-May-2026_dxyNySNf1d/*.csv"))
print(f"Found {len(csv_paths)} csv files")

dfs = []
for file in tqdm(csv_paths, desc="Loading CSV"):
    try:
        df = pd.read_csv(
            file,
            sep=",",
            engine="python",
            encoding="utf-8",
            quotechar='"',
            on_bad_lines="skip",
        )
        dfs.append(df)
    except Exception as e:
        print(f"[ERROR] {file}: {e}")

data = pd.concat(dfs, ignore_index=True)
print("Total rows:", len(data))
print(data.columns.tolist())
```

```python
# Bersihkan tanda kutip tunggal pada kolom string
for col in data.select_dtypes(include=["object"]).columns:
    data[col] = data[col].astype(str).str.replace("'", "", regex=False)

# Hapus kolom setelah kolom "label" (jika ada)
cols_to_keep = data.columns.tolist()
if "label" in cols_to_keep:
    label_index = cols_to_keep.index("label")
    cols_to_keep = cols_to_keep[: label_index + 1]
    data = data[cols_to_keep]
    print("Columns after dropping unnecessary ones:", data.columns.tolist())
else:
    print("'label' column not found. No columns were dropped.")

# Drop duplicates
data = data.drop_duplicates()
print("Total rows after dropping duplicates:", len(data))

data.to_csv("total_data_cleaned_komdigiSNf1d_final.csv", index=False, sep=",", encoding="utf-8")
print("Saved total_data_cleaned_komdigiSNf1d_final.csv")
```
